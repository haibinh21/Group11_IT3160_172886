## 2. Đề xuất project (W3)

### Bài toán
**Bối cảnh:** Dò mìn (Minesweeper) là trò chơi giải đố trên lưới ô cờ. Một số ô chứa mìn bị ẩn. Khi mở một ô an toàn, ô đó sẽ hiển thị một con số cho biết tổng số mìn nằm ở các ô lân cận (8 ô xung quanh). Mục tiêu của trò chơi là xác định (đánh cờ - flag) chính xác tất cả các ô chứa mìn mà không kích nổ bất kỳ quả mìn nào.

**Mục tiêu:** Xây dựng một tác tử thông minh (AI Agent) dựa trên tri thức (Knowledge-based Agent) có khả năng tự động phân tích các con số gợi ý trên bàn cờ, từ đó suy luận logic để tìm ra các ô chắc chắn an toàn (safe moves) hoặc chắc chắn chứa mìn (mines), đồng thời đưa ra quyết định di chuyển tối ưu hoặc ngẫu nhiên khi chưa đủ thông tin.


### Phương pháp dự kiến
Dự án sử dụng Tác tử dựa trên Tri thức (Knowledge-Based Agent) và Logic Mệnh đề (Propositional Logic):Biểu diễn tri thức (Knowledge Representation):Mỗi ô trên bàn cờ được xem như một biến mệnh đề (nhận giá trị True nếu chứa mìn, False nếu an toàn).Mối quan hệ giữa các ô được biểu diễn qua lớp Sentence dạng câu logic:$$\text{\{Tập hợp các ô lân cận\}} = \text{Count}$$(Ví dụ: $\{A, B, C\} = 1$ nghĩa là trong 3 ô A, B, C có chính xác 1 ô chứa mìn).Cơ chế suy luận logic (Inference Rules):Xác định ô an toàn / mìn trực tiếp:Nếu $\text{Count} = 0$, tất cả các ô trong tập hợp đều an toàn (mark_safe).Nếu $\text{Count} = \vert{}\text{Tập các ô}\vert{}$, tất cả các ô trong tập hợp đều là mìn (mark_mine).Suy luận tập con (Subset Inference):Nếu câu logic $S_1$ là tập con của câu logic $S_2$ ($S_1 \subset S_2$), ta có thể tạo ra câu logic mới:$$S_2 - S_1 = \text{Count}_2 - \text{Count}_1$$
Giúp thu hẹp không gian tìm kiếm và phát hiện ra các ẩn số mới.

Cấu trúc chương trình & Thuật toán quyết định (Decision Making):

Lớp Sentence: Quản lý tập hợp ô, số lượng mìn và các phương thức cập nhật khi phát hiện ô an toàn/mìn.

Lớp MinesweeperAI:

  - add_knowledge(cell, count): Cập nhật thông tin ô mới mở vào cơ sở tri thức (Knowledge Base) và kích hoạt quá trình suy luận lan truyền.

  - make_safe_move(): Chọn ô đã được xác nhận chắc chắn an toàn để mở.

  - make_random_move(): Khi không còn ô an toàn biết trước, AI chọn ngẫu nhiên một ô chưa mở và chưa bị đánh dấu mìn.

Lớp runner.py: Quản lý giao diện đồ họa (sử dụng thư viện pygame) để người dùng theo dõi AI thi đấu thực tế.
